## 1. Import the tools

Run the cell below by pressing the ▶ button or Shift+Enter.

```python
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
```

## 2. Load the Excel file

Upload `neurophysiology_plotting_data.xlsx` to JupyterLite, then run this cell.

The table contains one row for each recording and includes treatment group, stimulus intensity, response amplitude, and response latency.

```python
data = pd.read_excel('neurophysiology_plotting_data.xlsx')
data.head()
```

## 3. Inspect the dataset

How many observations and variables are present? What does each row represent? What are the units of the response variables?

```python
data.shape
```

```python
data.select_dtypes(include="number").describe()
```

## 4. Plot a relationship: scatter plot

Plot stimulus intensity against response amplitude. Add axis labels and a title.

**Question:** Does the response appear to change with stimulus intensity? Is a scatter plot appropriate for showing this relationship?

```python
plt.scatter(data['Stimulus intensity (mA)'], data['Response amplitude (mV)'])
plt.xlabel('Stimulus intensity (mA)')
plt.ylabel('Response amplitude (mV)')
plt.title('Response amplitude and stimulus intensity')
plt.show()
```

## 5. Compare groups: bar plot

Calculate the mean response amplitude for each treatment group and display it as a bar plot.

**Question:** What information is visible in the bar plot? What information is hidden? Why should error bars normally be considered when comparing groups?

```python
group_means = data.groupby('Treatment')['Response amplitude (mV)'].mean()
group_means
```

```python
group_means.plot(kind='bar', color=['steelblue', 'darkorange'])
plt.xlabel('Treatment')
plt.ylabel('Mean response amplitude (mV)')
plt.title('Mean response amplitude by treatment')
plt.xticks(rotation=0)
plt.show()
```

## 6. Show a distribution: histogram

Plot the distribution of response amplitudes.

**Question:** What does the histogram show that the bar plot does not? Describe the shape of the distribution.

```python
plt.hist(data['Response amplitude (mV)'], bins=8, edgecolor='black')
plt.xlabel('Response amplitude (mV)')
plt.ylabel('Number of recordings')
plt.title('Distribution of response amplitudes')
plt.show()
```
